```python
# If we are in Google Colab, mount Google Drive and install the package.

!pip install -e .. # Modify the path if necessary. It is NOT necessary to unzip the file.

import lls_sev_est as lls
import pandas as pd
```

```python
# Create DataFrame to save results

results = pd.DataFrame(columns=[
    "image_route","n_leaflets","nodes","missing_leaflets",
    "severity_%","defoliation_%","adjusted_severity_%"])
```

```python
# Image processing
# Modify with the name of our image and the number of nodes.

results, fig = lls.lls_sev_est(
    "IMG_20250424_0004.tif",
    nodes=10,
    results_df=results,
    show_results=True,
    verbose=True,
    return_figure=True
)
```

```python
# Show the resulting DataFrame
results
```

```python
# Save figures
lls.save_figure(fig, "outputs/leaf_severity", format="png", dpi=300)
```

```python
# Image processing without showing the results.
# Modify with the name of our image and the number of nodes.

results = lls.lls_sev_est(
    "IMG_20250424_0002.tif",
    nodes=10,
    results_df=results,
    show_results= False,
    verbose=True,
    return_figure=False
)
```

```python
# Show the resulting DataFrame
results
```

```python
# Save DataFrame
lls.save_results(results, "outputs/results.xlsx")
```
